#### **Creating DataFrame**

```python
import pandas as pd
import matplotlib.pyplot as plt

data={'Name':['Amit','Pranali','Sakshi',"Harshada",'Gauri','Raj','Rohan','Pooja','Karan','Anjali'],
      'Python':[85,92,76,60,88,72,95,68,89,78],
      'SQL':[82,89,74,58,91,77,94,66,88,80],
      'Maths':[82,89,74,58,91,77,94,66,88,80],
      'Attendance':[90,96,85,70,80,92,98,72,94,86]}
df=pd.DataFrame(data)
print(df)
```

#### **Average calculation**

```python
df['Average']=df[['Python','SQL','Maths']].mean(axis=1)
print(df)
```

#### **Highest Scoring Student**

```python
df.loc[df['Average'].idxmax()]
```

#### **Class Average Marks**

```python
class_avg=df['Average'].mean()
print("Class Average:",round(class_avg,2))
```

#### **Lowest Scoring Student**

```python
df.loc[df['Average'].idxmin()]
```

#### **Subject-wise Average Marks**

```python
df[['Python','SQL','Maths']].mean()
```

#### **Subject-wise Average Marks Visualization**

```python
sub_avg=df[['Python','SQl','Maths']].mean()
sub_avg.plot(kind='bar')
plt.title('Subject Wise Average Marks')
plt.xlabel('Subjects')
plt.ylabel('Average Marks')
plt.show()
```

```python
sub_avg=df[['Python','SQL','Maths']].mean()
sub_avg.plot(kind='bar')
plt.title('Subject Wise Average Marks')
plt.xlabel('Subjects')
plt.ylabel('Average Marks')
plt.show()
```

```python
print(df.columns.tolist())
```

#### **Student Wise Average marks**

```python
df.plot(x='Name',y='Average',kind='bar',color='teal',legend=False)
plt.title("Student Wise Average Marks")
plt.xlabel('Students')
plt.ylabel('Average Marks')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df.plot(x='Name',y='Average',kind='barh',color='purple',legend=False)
plt.title("Student Wise Average Marks")
plt.xlabel('Students')
plt.ylabel('Average Marks')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### **Attendence vs Average Marks**

```python
plt.scatter(df['Attendance'],df['Average'])
plt.title("Attendence vs Average Marks")
plt.xlabel('Attendences')
plt.ylabel('Average Marks')
plt.show()
```
